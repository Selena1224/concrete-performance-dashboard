```python
# Import the Pandas library and use "pd" as its abbreviation
import pandas as pd

# Load the concrete strength dataset from the data folder
# "../" means go up one folder from the notebooks folder
# Store the dataset in a DataFrame named "df"
df = pd.read_excel("../data/Concrete_Data.xls")

# Display the dataset dimensions (number of rows and columns)
print("Dataset dimensions:", df.shape)

# Display the first 5 rows to inspect the column names and values
display(df.head())
```

```python
# Check the data types and number of non-null values in each column
df.info()

# Check how many missing values each column has
print("\nMissing values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Get summary statistics for each column
# Includes the mean, standard deviation, minimum, maximum, and quartiles
display(df.describe())
```

```python
# Rename the columns so they're easier to work with
# Ingredient quantities are measured in kg/m³
# Age is measured in days, and strength in MPa

df.columns = [
    "cement",
    "slag",
    "fly_ash",
    "water",
    "superplasticizer",
    "coarse_aggregate",
    "fine_aggregate",
    "age_days",
    "strength_mpa"
]

# Check the updated column names
display(df.head())
```

```python
# Count rows that are exact duplicates, including strength
print("Exact duplicates:", df.duplicated().sum())

# Check whether the same ingredients and curing age
# appear more than once, even if the measured strength differs
mix_columns = df.columns.drop("strength_mpa").tolist()

repeated_mix = df.duplicated(subset=mix_columns, keep=False)

print("Rows sharing the same mix and curing age:", repeated_mix.sum())

# View some repeated observations to compare their strengths
display(
    df.loc[repeated_mix]
    .sort_values(mix_columns)
    .head(15)
)
```

```python
# Group rows that have identical values across all columns
# Count how many times each identical record appears
duplicate_groups = (
    df.groupby(df.columns.tolist(), dropna=False)
    .size()
    .reset_index(name="times_recorded")
)

# Keep only records that appear more than once
duplicate_groups = duplicate_groups[
    duplicate_groups["times_recorded"] > 1
]

# Sort so the most frequently repeated records appear first
duplicate_groups = duplicate_groups.sort_values(
    "times_recorded", ascending=False
)

# Check how many distinct groups of duplicates we have
print("Duplicate groups:", len(duplicate_groups))

# Count the extra copies beyond the first occurrence
print("Extra duplicate rows:",
      (duplicate_groups["times_recorded"] - 1).sum())

# Display the repeated records
display(duplicate_groups.head(10))
```

```python
# Keep a separate copy of the original data
df_original = df.copy()

# Remove rows where every column has the same value
# Keep the first occurrence of each record
df_clean = df.drop_duplicates().copy()

# Reset the row numbers after removing duplicates
df_clean = df_clean.reset_index(drop=True)

# Check how many records we removed
print("Original rows:", len(df_original))
print("Cleaned rows:", len(df_clean))
print("Rows removed:", len(df_original) - len(df_clean))

# Make sure there are no exact duplicates left
print("Remaining duplicates:", df_clean.duplicated().sum())

# Save the cleaned data so we can use it later
df_clean.to_csv("../data/concrete_clean.csv", index=False)
```

```python
# List the seven ingredients used in the concrete mixtures
ingredients = [
    "cement", "slag", "fly_ash", "water",
    "superplasticizer", "coarse_aggregate", "fine_aggregate"
]

# Check for negative ingredient quantities
print("Rows with negative ingredient values:",
      (df_clean[ingredients] < 0).any(axis=1).sum())

# Check for invalid curing ages
print("Rows with zero or negative curing age:",
      (df_clean["age_days"] <= 0).sum())

# Check for invalid compressive strength values
print("Rows with zero or negative strength:",
      (df_clean["strength_mpa"] <= 0).sum())

# Check whether any mixtures have no cement, slag, or fly ash
# These three materials make up the binder in our dataset
binder = df_clean[["cement", "slag", "fly_ash"]].sum(axis=1)

print("Rows with no binder:", (binder <= 0).sum())

# Review the minimum, maximum, and distribution of each variable
# The 1st and 99th percentiles help us spot unusually low or high values
display(
    df_clean.describe(percentiles=[0.01, 0.25, 0.5, 0.75, 0.99])
    .round(2)
    .T
)
```

```python
# Create a working copy so the cleaned dataset stays unchanged
df_model = df_clean.copy()

# Calculate the total binder content (cement + slag + fly ash)
# All three ingredients are measured in kg/m³
df_model["total_binder"] = (
    df_model["cement"] +
    df_model["slag"] +
    df_model["fly_ash"]
)

# Calculate the water-to-binder ratio
# This compares the amount of water with the total binder content
df_model["water_binder_ratio"] = (
    df_model["water"] / df_model["total_binder"]
)

# Calculate the percentage of binder made up of slag and fly ash
# This helps us compare different cement replacement levels
df_model["cement_replacement_pct"] = (
    (df_model["slag"] + df_model["fly_ash"]) /
    df_model["total_binder"]
) * 100

# Review the new variables and their distributions
display(
    df_model[
        ["total_binder", "water_binder_ratio",
         "cement_replacement_pct"]
    ].describe().round(2)
)

# Check whether the calculations created any missing values
print("\nMissing values in new variables:")
print(
    df_model[
        ["total_binder", "water_binder_ratio",
         "cement_replacement_pct"]
    ].isnull().sum()
)
```

```python
# Count the number of unique concrete recipes
# A recipe is defined by its seven ingredient quantities
# Curing age is excluded because one recipe can be tested at different ages
unique_mixes = df_model[ingredients].drop_duplicates().shape[0]

print("Total observations:", len(df_model))
print("Unique concrete recipes:", unique_mixes)

# Check how many recipes were tested at multiple curing ages
# This will help us decide how to split the data for modeling
mix_age_counts = (
    df_model.groupby(ingredients)["age_days"]
    .nunique()
)

print("Recipes tested at multiple ages:",
      (mix_age_counts > 1).sum())

# Save the prepared dataset with our three additional variables
# Keep the original and previously cleaned datasets unchanged
df_model.to_csv("../data/concrete_prepared.csv", index=False)

print("\nPrepared dataset saved successfully!")
```
